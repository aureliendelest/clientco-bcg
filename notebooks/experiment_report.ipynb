# Rapport des expériences churn

Ce carnet ne lit **que** `results/` (produit par `python -m churn collect`) et génère tous les graphiques et tableaux du rapport. Aucune expérience n'est lancée ici.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RESULTS = ROOT / "results"


def load(name: str) -> pd.DataFrame:
    """Read one table of results/, or an empty frame if it has not been collected yet."""
    path = RESULTS / f"{name}.parquet"
    return pd.read_parquet(path) if path.exists() else pd.DataFrame()


runs = load("runs")
metrics = load("metrics")
predictions = load("predictions")
importances = load("importances")
print({k: len(v) for k, v in dict(runs=runs, metrics=metrics, predictions=predictions, importances=importances).items()})
```
